### Operações em pandas

```python
print("Select e Where")

filtro = df2['departamento'] == 'Engenharia'
df_engenharia = df2.loc[filtro, ['nome', 'salario']]
print(df_engenharia)
```

```python
print("GROUP BY E AVG agupamento e calcular a média")

media_salarial = df2.groupby('departamento')['salario'].mean()
print(media_salarial)
```

```python
print("ORDER BY - Ordenação")
df_ordenado = df2.sort_values(by='salario', ascending=False)
print(df_ordenado[['nome', 'salario']])
```

```python
print("ALTER TABLE E ADD COLUMN")

df2['salario_com_bonus'] = df2['salario'] * 1.10
print(df2[['nome', 'salario', 'salario_com_bonus',]].head())
```

### Debugger

```python
def calcular_media_idades(lista_idades):
    soma = 0
    for idade in lista_idades:
        if type(idade) == str:
            idade = int(idade)
        soma += idade

    if not len(lista_idades):
        return 0

    media = soma / len(lista_idades)
    return media

# cenario que funciona perfeito
# idades_equipe = [25, 30, 22, 27]
# print(f"Média 1: {calcular_media_idades(idades_equipe)}")

# cenario 2
idades_sem_ninguem = [25, 30, '22', 27]
print(f"Média 2: {calcular_media_idades(idades_sem_ninguem)}")
```

### Classes

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def exibir_info(self):
        status = "Adulto" if self.age >= 10 else "Menor"
        print(f"Usuário: {self.name} | Idade: {self.age} | Status{status}")

# Instanciando objetos na memória
user1 = User("Duarte Junior", 27)
user2 = User("Ana Silva", 16)

# chamando os metodos

print("Sistemas de usuário")
user1.exibir_info()
user2.exibir_info()
```

```python
class User:
    def __init__(self, name, age, password):
        # publicos
        self.name = name
        self.age = age

        # privados
        self.__password = password

    def exibir_perfil(self):
        print(f"Perfil: Nome {self.name} | Idade: {self.age}")

    def verificar_senha(self, tentativa_senha):
        if tentativa_senha == self.__password:
            return "Acesso liberado!"
        else:
            return "Senha incorreta!"

professor = User("Duarte Junior", 27, "senha_secreta")
aluno = User("joão Silva", 20, "123456")

print("Sistemas de usuarios")
professor.exibir_perfil()
aluno.exibir_perfil()

print("Teste de encapsulamento")
print("Tentativa de login do João:")
print(aluno.verificar_senha("senha errada"))
print(aluno.verificar_senha("123456"))
```
