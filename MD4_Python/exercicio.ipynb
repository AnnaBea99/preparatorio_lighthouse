```python
import pandas as pd
```

```python
# carregando arquivo csv
order_details = pd.read_csv('order_details.csv')

# carregando arquivo json 
orders = pd.read_json('public-orders.jsonl', lines=True)
```

```python
pd.DataFrame(order_details)
print(order_details.head(5))
print(order_details.info())
```

```python
# Realizando um join
merged = pd.merge(order_details, orders, on='order_id', how='inner')

df1 = pd.DataFrame(merged[['order_id', 'unit_price']])
df1.head(5)
```

```python
# realizando a ordenação pelo preco unitario

preco_unitario = merged['unit_price']
filtrado = merged[preco_unitario > 100]
filtrado = filtrado.sort_values(by='unit_price', ascending=False)

pd.DataFrame(filtrado.head(5))
```

```python
# Pegando os volores unicos dos consumidores pelo id
customer_ids = filtrado['customer_id'].unique()

# exportando para um novo csv
pd.DataFrame({'customer_id': customer_ids}).to_csv('customers_above_100.csv', index=False)
```
