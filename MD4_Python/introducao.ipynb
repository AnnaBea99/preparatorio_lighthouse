### String

```python
frase = "isso aqui é uma frase"
nova_frase = ["Maçã", "banana", "torta"]

print(str.capitalize(frase)) # primeira letra maiuscula
print(str.upper(frase)) # tudo em caps
print(str.lower(frase)) # tudo em minusculo 
print(frase.split(' ', 1)) # separa de acordo com o separador escolhido
print(frase.split()) # separa de acordo com o separador escolhido

juncao = ", ".join(nova_frase) # juntar itens ou palavras
print(juncao)

print(frase.count("a")) # retorna a quantidade de vezes que o valor aparece

print(frase.replace("a", "o", 1)) # substituir um valor
```

### Sequence

#### tupla ()
mutável, possuem index

```python
tupla = (1,2,3,4,2,4,7,8,9,9,5,6,7,2)
tupla2 = ("Maçã", "banana", "torta")
letras = tuple("banana")

# metodos
print("conta a qnt de itens")
print(letras)
print(letras.count("a"))
print(tupla.count(2))
print(tupla2.count("banana"))

print()

print("indica a posição")
print(tupla.index(9))
print(tupla2.index("torta"))
print(letras.index("a"))

print()

print("mostra o tamanho")
print(len(tupla))
print(len(tupla2))
print(len(letras))

print()

print("acessando tupla")

print(tupla2[0]) 
print(tupla2[1])  
```

#### Listas []

- odernáveis, mutáveis e podem ser duplicadas, possuem index


```python
lista = ["banana", "torta", 1999, "abacaxi", 27, 17]
lista2 = ["Banana", "torta", "Abacaxi", "Uva", "Torta de maçã"]

print("acessando lista: ", lista[2])

print("adicionando item: ", lista.append("banana"))
print("lista atualizada: ", lista)

print("contando:", lista.count("banana"))

print("index: ", lista.index(27))

print("inserir em uma determinda posição: ", lista.insert(1, "torta de maçã"))
print(lista)

lista[0] = "uva"
print("inserir novo valor: ", lista[0])


print("remove um item na posição escolhida: ", lista.pop(5))
print(lista)

print("removendo outro item: ", lista.remove("torta"))
print(lista)

lista2.sort()
print("ordena a lista: ", lista2)

print("mostra o tamanho da lista: ", len(lista))
print("mostra o tamanho da lista2: ", len(lista2))
```

### Condicionais

```python
num1 = 9
num2 = 12

if num1 > num2:
    print(f"{num1} é maior e {num2} é menor")
elif num1 == num2:
    print("números iguais")
else:
    print(f"{num2} é maior e {num1} é menor")
```

### Operators

```python
numeros = [2,3,7,9,34,55]
copia = numeros
lista_nova = [2,3,7,9,34,55]
x = 3
y = 12

# And, Or, Not

print(x > 1 and y > 5) # True

print(x == 9 or y == 12) # True

print(not (x==3)) # x é igual a 3 mas com o not ali o True vira False


# ==================================
# Is, Is not

# is checa se tem a mesma identidade/ endereço de memória
print(numeros is copia) # true

print(numeros is lista_nova) # false

# is not verifica se não são o mesmo objeto
print(numeros is not lista_nova) # true

# =======================================
# in not in

# in verifica se o elemento esta na sequencia
print(3 in numeros) # true


# not in verifica se o elemento não esta na sequencia
print(99 not in numeros) # true
```

### Loops

```python
# for percorre o que é passado

texto = "banana"

for i in texto:
    print(f"letra: {i}")
```

```python
# range(start, stop, step)

for numero in range(1, 11): 
    print(f"Tabuada do {numero}")
    for i in range(1,11):
        print(f"{numero} x {i} = {numero * i}")
    print()
```

```python
# usando loops com dicionario

precos = {
    "Torta de maça": 10,
    "Torta de chocolate": 15
}

#lendo a chave e valor juntos
for key, value in precos.items():
    print(f"O preço de {key} é R${value}")

print()
print("=========================================")
print()

# lendo só as chaves
for key in precos:
    print(f"Torta disponível {key} custa R${precos[key]}")
```

```python
# Enumerate
# Entrega o índice e o item

cores = ["Rosa", "Roxo"]

for i, valor in enumerate(cores):
    print(f"indice: {i} Cor: {valor}")
```

```python
# While

i = 1

while i <= 10:
    print(f"Tabuada do {i}")
    for x in range(1,11):
        print(f"{i} x {x} = {i * x}") 

    print()
    i += 1
```

```python
i = 0
while i < 5:
    i += 1  
    
    if i == 2:
        print("Pulei o 2!")
        continue  
        
    if i == 4:
        print("Interrompendo no 4!")
        break 
        
    print(f"Contagem: {i}")
```
